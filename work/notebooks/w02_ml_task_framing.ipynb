## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/f230865-netizen/FlyRankAI-Assignment-1/main/data/raw/content_refresh_anonymized.csv")

feature_cols = ['avg_position','position_tier','word_count','word_count_tier',
                 'content_type','main_intent','freshness_tier','age_tier',
                 'competition_level','search_volume','content_age_days']
target_source = 'trend_direction'

lane_slice = df[['content_id'] + feature_cols + [target_source]].copy()
lane_slice['is_declining'] = (lane_slice[target_source] == 'down').astype(int)

print("One row =", "one content page")
print("Shape:", lane_slice.shape)
lane_slice.head(5)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
print(lane_slice.groupby('is_declining')['avg_position'].mean())
```
